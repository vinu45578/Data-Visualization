```python
wins = df['match_winner'].value_counts()
```

```python
sns.barplot(y=wins.index,x=wins.values,palette='rainbow')
plt.title("Most Wins of Teams")
plt.xlabel('Matches Counts')
plt.ylabel('Teams')
```

```python

sns.countplot(data=df, x='toss_decision', palette='muted',)
plt.title('Toss Decisions Trends',loc='center')
plt.xlabel('Decision')
plt.ylabel('Count')
```

```python
count=df[df['toss_winner']==df['match_winner']]['match_id'].count()
percentage=(count/df.shape[0])*100
print(percentage.round(2))
```

```python
sns.countplot(x=df['won_by'],legend="auto")
plt.title('Won By')
```

```python
sns.countplot(data=df, x='won_by', hue='toss_decision')
plt.title('Won By vs Toss Decision')
plt.show()
```

```python
count=df['player_of_the_match'].value_counts().head(10)
count
```

```python
top_score=df.groupby('top_scorer')['highscore'].sum().sort_values(ascending=False).head()
```

```python
df['wickets_taken']=df['best_bowling_figure'].apply(lambda x: x.split('--')[0])
```

```python
df
```

```python
df['wickets_taken']=df['wickets_taken'].astype(int)
```

```python
best_bowl = df.groupby('best_bowling')['wickets_taken'].sum().sort_values(ascending=False).head(10)
```

```python
best_bowl
```

```python
best_bowl.plot(kind='bar',color='gold')
plt.title('Best Bowling Figures')
plt.xlabel('Count')
plt.ylabel('Bowlers')
plt.show()
```
